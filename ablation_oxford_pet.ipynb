```python
# ==========================================================
# COMPLETE ABLATION STUDY FOR DP-GCViT
# Oxford-IIIT Pet Dataset | 37 Breeds
# ==========================================================

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Install required packages
!pip install timm -q  # <-- FIXED: timm instead of timq

# Imports
import os
import copy
import random
import numpy as np
import pandas as pd
from tqdm import tqdm
from sklearn.metrics import precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.optim as optim
from torch.cuda.amp import autocast, GradScaler
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from PIL import Image

import timm

# ==========================================================
# PATHS (UPDATE THESE IF NEEDED)
# ==========================================================

BASE_SAVE_DIR = "/content/drive/MyDrive/dpGCViT_new_Results"
ABLATION_DIR = os.path.join(BASE_SAVE_DIR, "ablation_study")
os.makedirs(ABLATION_DIR, exist_ok=True)

DATA_ROOT = "/content/drive/MyDrive/The Oxford-IIIT Pet Dataset"
IMAGES_DIR = os.path.join(DATA_ROOT, "images", "images")
ANNOT_DIR = os.path.join(DATA_ROOT, "annotations", "annotations")
TRAINVAL_FILE = os.path.join(ANNOT_DIR, "trainval.txt")
TEST_FILE = os.path.join(ANNOT_DIR, "test.txt")

# ==========================================================
# CONFIGURATION
# ==========================================================

ABLATION_EPOCHS = 20
ABLATION_PATIENCE = 5
BATCH_SIZE = 8
ACCUM_STEPS = 2
IMG_SIZE = 224

# ==========================================================
# REPRODUCIBILITY
# ==========================================================

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ==========================================================
# DATASET CLASS
# ==========================================================

class OxfordPetDataset(Dataset):
    def __init__(self, list_file, img_dir, transform=None):
        self.transform = transform
        self.img_dir = img_dir
        self.samples = []

        # Get all classes first
        all_classes = set()
        with open(list_file) as f:
            for line in f:
                name = line.strip().split()[0]
                breed = name.rsplit("_", 1)[0]
                all_classes.add(breed)

        self.classes = sorted(list(all_classes))
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}

        # Load samples
        with open(list_file) as f:
            for line in f:
                name = line.strip().split()[0]
                breed = name.rsplit("_", 1)[0]
                self.samples.append((name + ".jpg", self.class_to_idx[breed]))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_name, label = self.samples[idx]
        image = Image.open(os.path.join(self.img_dir, img_name)).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, label

# ==========================================================
# TRANSFORMS
# ==========================================================

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.6, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(0.3, 0.3, 0.3, 0.1),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.4)
])

eval_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

# ==========================================================
# ABLATION CONFIGURATIONS
# ==========================================================

ablation_configs = [
    {"name": "Full DP-GCViT",    "part_fc": True,  "global_fc": True,  "gate": True},
    {"name": "No Gate",          "part_fc": True,  "global_fc": True,  "gate": False},
    {"name": "Part Only",        "part_fc": True,  "global_fc": False, "gate": False},
    {"name": "Global Only",      "part_fc": False, "global_fc": True,  "gate": False},
]

# ==========================================================
# MODEL
# ==========================================================

class DP_GCViT_Ablation(nn.Module):
    def __init__(self, backbone, num_classes,
                 use_part_fc=True, use_global_fc=True, use_gate=True):
        super().__init__()

        self.backbone = backbone
        self.num_features = backbone.num_features
        self.use_part_fc = use_part_fc
        self.use_global_fc = use_global_fc
        self.use_gate = use_gate

        self.proj = nn.Sequential(
            nn.LayerNorm(self.num_features),
            nn.Linear(self.num_features, self.num_features),
            nn.GELU()
        )

        self.part_fc = nn.Sequential(
            nn.Dropout(0.3),
            nn.Linear(self.num_features, num_classes)
        )
        self.global_fc = nn.Sequential(
            nn.Dropout(0.3),
            nn.Linear(self.num_features, num_classes)
        )

        self.gate = nn.Sequential(
            nn.Linear(self.num_features * 2, self.num_features),
            nn.GELU(),
            nn.Linear(self.num_features, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        feat = self.backbone.forward_features(x)
        if feat.dim() == 4:
            feat = feat.flatten(2).transpose(1, 2)

        global_feat = feat.mean(dim=1)
        part_feat = feat.max(dim=1)[0]

        global_feat = self.proj(global_feat)
        part_feat = self.proj(part_feat)

        if self.use_part_fc:
            part_out = self.part_fc(part_feat)
        if self.use_global_fc:
            global_out = self.global_fc(global_feat)

        if self.use_part_fc and self.use_global_fc:
            if self.use_gate:
                alpha = self.gate(torch.cat([global_feat, part_feat], dim=1))
                return alpha * part_out + (1 - alpha) * global_out
            else:
                return 0.5 * part_out + 0.5 * global_out
        elif self.use_part_fc:
            return part_out
        else:
            return global_out

# ==========================================================
# TRAINING FUNCTION
# ==========================================================

def train_ablation(model, train_loader, val_loader, epochs, patience):
    optimizer = optim.AdamW(model.parameters(), lr=3e-4, weight_decay=0.05)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    scaler = GradScaler()

    best_acc = 0
    best_model = copy.deepcopy(model)
    patience_counter = 0

    for epoch in range(epochs):
        # Training
        model.train()
        correct, total = 0, 0
        optimizer.zero_grad()

        for i, (x, y) in enumerate(tqdm(train_loader, desc=f"Epoch {epoch+1}")):
            x, y = x.to(device), y.to(device)

            with autocast():
                out = model(x)
                loss = criterion(out, y) / ACCUM_STEPS

            scaler.scale(loss).backward()

            if (i + 1) % ACCUM_STEPS == 0:
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()

            pred = out.argmax(1)
            correct += (pred == y).sum().item()
            total += y.size(0)

        train_acc = 100 * correct / total

        # Validation
        model.eval()
        correct, total = 0, 0
        with torch.no_grad():
            for x, y in val_loader:
                x, y = x.to(device), y.to(device)
                out = model(x)
                pred = out.argmax(1)
                correct += (pred == y).sum().item()
                total += y.size(0)

        val_acc = 100 * correct / total
        scheduler.step()

        print(f"Epoch {epoch+1}: Train Acc={train_acc:.2f}%, Val Acc={val_acc:.2f}%")

        # Early stopping
        if val_acc > best_acc:
            best_acc = val_acc
            best_model = copy.deepcopy(model)
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                break

    return best_model, best_acc

# ==========================================================
# TEST FUNCTION
# ==========================================================

def test_model(model, test_loader):
    model.eval()
    y_true, y_pred = [], []

    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)

            # TTA: original + flipped
            out1 = model(x)
            out2 = model(torch.flip(x, [3]))
            out = (out1 + out2) / 2

            pred = out.argmax(1)
            y_true.extend(y.cpu().numpy())
            y_pred.extend(pred.cpu().numpy())

    top1 = 100 * np.mean(np.array(y_true) == np.array(y_pred))
    precision = 100 * precision_score(y_true, y_pred, average='macro', zero_division=0)
    recall = 100 * recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = 100 * f1_score(y_true, y_pred, average='macro', zero_division=0)

    return top1, precision, recall, f1

# ==========================================================
# LOAD DATA
# ==========================================================

print("=" * 70)
print("LOADING DATASET")
print("=" * 70)

# Create full dataset to get number of classes
temp_dataset = OxfordPetDataset(TRAINVAL_FILE, IMAGES_DIR, eval_transform)
num_classes = len(temp_dataset.classes)
print(f"Number of classes: {num_classes}")

# Create train/val split
full_dataset = OxfordPetDataset(TRAINVAL_FILE, IMAGES_DIR, eval_transform)
indices = list(range(len(full_dataset)))
random.shuffle(indices)
split = int(0.8 * len(indices))
train_idx, val_idx = indices[:split], indices[split:]

# Create datasets with proper transforms
train_dataset = Subset(
    OxfordPetDataset(TRAINVAL_FILE, IMAGES_DIR, train_transform),
    train_idx
)
val_dataset = Subset(
    OxfordPetDataset(TRAINVAL_FILE, IMAGES_DIR, eval_transform),
    val_idx
)
test_dataset = OxfordPetDataset(TEST_FILE, IMAGES_DIR, eval_transform)

# Create data loaders
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

print(f"Train samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"Test samples: {len(test_dataset)}")

# ==========================================================
# RUN ABLATION STUDY
# ==========================================================

print("\n" + "=" * 70)
print("STARTING ABLATION STUDY")
print("=" * 70)

results = []

for cfg in ablation_configs:
    print("\n" + "=" * 70)
    print(f"RUNNING: {cfg['name']}")
    print("=" * 70)

    # Create model
    backbone = timm.create_model("gcvit_tiny", pretrained=True, num_classes=0)
    model = DP_GCViT_Ablation(
        backbone,
        num_classes,
        use_part_fc=cfg["part_fc"],
        use_global_fc=cfg["global_fc"],
        use_gate=cfg["gate"]
    ).to(device)

    print(f"Total parameters: {sum(p.numel() for p in model.parameters()):,}")

    # Train
    best_model, best_val_acc = train_ablation(
        model, train_loader, val_loader,
        epochs=ABLATION_EPOCHS, patience=ABLATION_PATIENCE
    )

    # Test
    top1, precision, recall, f1 = test_model(best_model, test_loader)

    results.append({
        "Configuration": cfg["name"],
        "Best Val Acc (%)": round(best_val_acc, 2),
        "Test Top-1 (%)": round(top1, 2),
        "Test Precision (%)": round(precision, 2),
        "Test Recall (%)": round(recall, 2),
        "Test F1 (%)": round(f1, 2)
    })

    print(f"\n{cfg['name']} Results:")
    print(f"  Best Val Acc: {best_val_acc:.2f}%")
    print(f"  Test Top-1: {top1:.2f}%")
    print(f"  Test F1: {f1:.2f}%")

    # Clear GPU memory
    del model, best_model, backbone
    torch.cuda.empty_cache()

# ==========================================================
# SAVE RESULTS
# ==========================================================

df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("FINAL ABLATION RESULTS")
print("=" * 70)
print(df.to_string(index=False))

# Save to CSV
csv_path = os.path.join(ABLATION_DIR, "ablation_results.csv")
df.to_csv(csv_path, index=False)
print(f"\nResults saved to: {csv_path}")

# Create bar plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Test Top-1 and F1
x = np.arange(len(df))
width = 0.35

ax1.bar(x - width/2, df['Test Top-1 (%)'], width, label='Top-1 Accuracy', color='steelblue')
ax1.bar(x + width/2, df['Test F1 (%)'], width, label='F1 Score', color='darkorange')
ax1.set_xlabel('Configuration')
ax1.set_ylabel('Score (%)')
ax1.set_title('Test Performance Comparison')
ax1.set_xticks(x)
ax1.set_xticklabels(df['Configuration'], rotation=45, ha='right')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: All metrics
metrics = ['Test Top-1 (%)', 'Test Precision (%)', 'Test Recall (%)', 'Test F1 (%)']
x_pos = np.arange(len(metrics))

for i, row in df.iterrows():
    values = [row[m] for m in metrics]
    ax2.bar(x_pos + i*0.2, values, width=0.2, label=row['Configuration'])

ax2.set_xlabel('Metrics')
ax2.set_ylabel('Score (%)')
ax2.set_title('Detailed Metrics by Configuration')
ax2.set_xticks(x_pos + 0.3)
ax2.set_xticklabels(metrics, rotation=45, ha='right')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plot_path = os.path.join(ABLATION_DIR, "ablation_results.png")
plt.savefig(plot_path, dpi=300)
plt.show()

print(f"\nPlot saved to: {plot_path}")

# Display final summary
print("\n" + "=" * 70)
print("SUMMARY - BEST PERFORMING CONFIGURATION")
print("=" * 70)
best_idx = df['Test Top-1 (%)'].idxmax()
best_config = df.iloc[best_idx]
print(f"Best Configuration: {best_config['Configuration']}")
print(f"Test Top-1 Accuracy: {best_config['Test Top-1 (%)']}%")
print(f"Test F1 Score: {best_config['Test F1 (%)']}%")
```
